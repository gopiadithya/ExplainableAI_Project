# 04 - Model Evaluation & Selection

Comprehensive evaluation of all models on the held-out test set:
- Accuracy, Precision, Recall, F1-Score
- ROC-AUC and PR-AUC
- Confusion matrices
- Weighted model selection prioritizing Recall for failure detection

```python
import sys
sys.path.insert(0, "..")
import pandas as pd
import joblib
from src.evaluate import evaluate_all_models, create_comparison_table, select_best_model

models = {
    "Logistic Regression": joblib.load("../models/logistic_regression.joblib"),
    "Random Forest": joblib.load("../models/random_forest.joblib"),
    "XGBoost": joblib.load("../models/xgboost.joblib")
}
X_test = pd.read_csv("../data/processed/X_test.csv")
y_test = pd.read_csv("../data/processed/y_test.csv").squeeze()

results = evaluate_all_models(models, X_test, y_test)
comp_table = create_comparison_table(results)
print(comp_table)

best_name, best_model, reasoning = select_best_model(results)
print(f"\nBest Model: {best_name}")
print(f"Reasoning: {reasoning}")
```
